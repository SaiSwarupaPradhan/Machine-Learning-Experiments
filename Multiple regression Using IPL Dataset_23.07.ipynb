```python
import pandas as pd

# Read the CSV file
df = pd.read_csv('IPL IMB381IPL2013.csv')

# Show first 5 rows
df.head()
```

```python
# Show all column names
print(df.columns)
```

```python
#The value which is to be predicted
y = df['SOLD PRICE']
```

```python
# Selecting feature columns (X)
X = df[['AGE', 'T-RUNS', 'T-WKTS', 'ODI-RUNS-S', 'ODI-SR-B',
        'ODI-WKTS', 'ODI-SR-BL', 'CAPTAINCY EXP', 'RUNS-S',
        'HS', 'AVE', 'SR-B', 'SIXERS', 'RUNS-C', 'WKTS',
        'AVE-BL', 'ECON', 'SR-BL', 'BASE PRICE']]

# Target column (y)
y = df['SOLD PRICE']

# Check the shape
print('X shape:', X.shape)
print('y shape:', y.shape)
```

```python
#Train and Test Data Split
from sklearn.model_selection import train_test_split

# 70% train, 30% test
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.30,     # 30% for testing
    random_state=42
)

# Check the sizes
print('X_train:', X_train.shape)
print('X_test :', X_test.shape)
print('y_train:', y_train.shape)
print('y_test :', y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression

# Create the model
model = LinearRegression()

# Train the model
model.fit(X_train, y_train)

print('Model training completed!')

# Number of coefficients learned
print('Number of coefficients:', len(model.coef_))

# Show first few coefficients
print(model.coef_[:5])
```

```python
# Predict SOLD PRICE for test players
y_pred = model.predict(X_test)

# Show first 5 predictions
print(y_pred[:5])
```

```python
comparison = pd.DataFrame({
    'Actual': y_test.values,
    'Predicted': y_pred
})

comparison.head(10)
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(y_test, y_pred)

print('R² Score:', r2)
```

```python
#Calculating Predictipon Error
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import numpy as np

# R2 Score
r2 = r2_score(y_test, y_pred)

# Error metrics
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)

print('R² Score:', r2)
print('MAE :', mae)
print('RMSE:', rmse)
```
