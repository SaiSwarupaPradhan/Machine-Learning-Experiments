```python

import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.preprocessing import MaxAbsScaler, RobustScaler, Normalizer

df = pd.read_csv("winequality-red.csv", sep=",")

print("Original Dataset:")
display(df.head())

# 1. Min-Max Normalization
df_minmax = pd.DataFrame(
    MinMaxScaler().fit_transform(df), columns=df.columns)
print("\nMin-Max Normalization:")
display(df_minmax.head())

# 2. Z-Score Standardization
df_zscore = pd.DataFrame(
    StandardScaler().fit_transform(df), columns=df.columns)
print("\nZ-Score Standardization:")
display(df_zscore.head())

# 3. Decimal Scaling
max_values = df.abs().max()
powers = np.ceil(np.log10(max_values.replace(0, 1)))
df_decimal = df / (10 ** powers)
print("\nDecimal Scaling:")
display(df_decimal.head())

# 4. Max-Abs Scaling
df_maxabs = pd.DataFrame(
    MaxAbsScaler().fit_transform(df), columns=df.columns)
print("\nMax-Abs Scaling:")
display(df_maxabs.head())

# 5. Robust Scaling
df_robust = pd.DataFrame(
    RobustScaler().fit_transform(df), columns=df.columns)
print("\nRobust Scaling:")
display(df_robust.head())

# 6. Unit Vector Normalization
df_unit = pd.DataFrame(
    Normalizer().fit_transform(df), columns=df.columns)
print("\nUnit Vector Normalization:")
display(df_unit.head())
```
